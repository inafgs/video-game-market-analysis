Se observó que en los años más antiguos el número de juegos lanzados es considerablemente bajo en comparación con los años más recientes.
````python
#Calcular ventas totales por plataforma

platform_sales = games.groupby('platform')['total_sales'].sum().sort_values(ascending=False)

print("Top 10 plataformas por ventas totales:")
print(platform_sales.head(10))

#Visualizado
plt.figure(figsize=(12, 6))
platform_sales.head(10).plot(kind='bar')
plt.title('Top 10 plataformas por ventas totales')
plt.xlabel('Platform')
plt.ylabel('Ventas totales (millones)')
plt.xticks(rotation=45)
plt.show()
````

````python
#Top platforms
top_platforms = platform_sales.head(8).index.tolist()

#tabla de ventas por año y plataforma
platform_year_sales = games[games['platform'].isin(top_platforms)].groupby(['year_of_release', 'platform'])['total_sales'].sum().unstack(fill_value=0)

#Visualizar evolución
plt.figure(figsize=(15, 8))
for platform in top_platforms:
    if platform in platform_year_sales.columns:
        plt.plot(platform_year_sales.index, platform_year_sales[platform],
                marker=0, label=platform, linewidth=2)

plt.title('Evolución de ventas por plataforma a lo largo del tiempo')
plt.xlabel('Año')
plt.ylabel('Ventas totales (millones)')
plt.legend()
plt.grid(True, alpha=0.30)
plt.show()
````

<div class="alert alert-block alert-success">
<b>Comentario de Revisor            </b> <a class="tocSkip"></a>

Excelente, muy buen gráfico, permite apreciar el periodo de vida de cada consola, incluyendo su auge y luego caída.

</div>

````python
# Últimos años de actividad por plataforma
platform_last_year = games.groupby('platform')['year_of_release'].max().sort_values(ascending=False)

print("Últimos años de actividad por plataforma:")
print(platform_last_year.head(15))

# Plataformas que no tienen ventas recientes
extinct_platforms = platform_last_year[platform_last_year < 2010]
print(f"\nPlataformas 'extintas' (sin actividad después de 2010): {len(extinct_platforms)}")
print(extinct_platforms.head(10))
````

````python
# Calcular duración de cada plataforma
platform_lifespan = games.groupby('platform')['year_of_release'].agg(['min', 'max'])
platform_lifespan['lifespan'] = platform_lifespan['max'] - platform_lifespan['min']
platform_lifespan = platform_lifespan.sort_values(by='lifespan', ascending=False)

print("Duración de vida de las plataformas (años):")
print(platform_lifespan.head(10))
````

````python
#¿Qué plataformas son líderes en ventas en el período relevante (2011 en adelante)?
recent_games.groupby('platform')['total_sales'].sum().sort_values(ascending=False).head()
````

````python
recent_top_platforms = (
    recent_games
    .groupby('platform')['total_sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)
    .index
    .tolist()
)

print(recent_top_platforms)

recent_top_platforms_sales = (
    recent_games[recent_games['platform'].isin(recent_top_platforms)]
    .groupby(['year_of_release', 'platform'])['total_sales']
    .sum()
    .unstack(fill_value=0)
)
````

````python
plt.figure(figsize=(12, 6))

for platform in recent_top_platforms_sales.columns:
    plt.plot(
        recent_top_platforms_sales.index,
        recent_top_platforms_sales[platform],
        marker='o',
        label=platform,
        linewidth=2
    )

plt.title('Plataformas líderes en ventas')
plt.xlabel('Año')
plt.ylabel('Ventas totales (millones)')
plt.legend()
plt.grid(True, alpha=0.30)
plt.show()
````

El análisis de la relación entre las calificaciones de los críticos (critic_score) y las ventas totales de los juegos en la plataforma PS4 muestra que existe una correlación positiva entre ambas variables. Esto sugiere que, aunque las reseñas profesionales pueden influir en el rendimiento comercial de un juego, no son el único factor determinante en su éxito.

En consecuencia, se puede afirmar que las críticas de expertos tienen cierto impacto en las ventas de los juegos en la plataforma PS4, pero otros factores como la popularidad de la franquicia, el marketing o el género del juego también pueden desempeñar un papel importante en su desempeño en el mercado.

````python
# 1. Preparación de datos (filtrar por plataforma y eliminar nulos en user_score)
xbox_one_scores = games[(games['platform'] == 'XOne') & (games['user_score'].notna())]['user_score']
pc_scores = games[(games['platform'] == 'PC') & (games['user_score'].notna())]['user_score']

# 2. Prueba de Levene para verificar si las varianzas son iguales
levene_res = stats.levene(xbox_one_scores, pc_scores)

# 3. Realizar prueba t
# Si p-value < 0.05, las varianzas son diferentes
results_platforms = stats.ttest_ind(xbox_one_scores, pc_scores, equal_var=(levene_res.pvalue > 0.05))

print(f"P-value para plataformas: {results_platforms.pvalue:.4f}")

if results_platforms.pvalue < alpha:
    print("Rechazamos la hipótesis nula: las calificaciones promedio son diferentes.")
else:
    print("No podemos rechazar la hipótesis nula: no hay evidencia de una diferencia significativa.")
````

````python
# 1. Preparación de datos
action_scores = games[(games['genre'] == 'Action') & (games['user_score'].notna())]['user_score']
sports_scores = games[(games['genre'] == 'Sports') & (games['user_score'].notna())]['user_score']

# 2. Prueba de Levene
levene_genres = stats.levene(action_scores, sports_scores)

# 3. Realizar prueba t
results_genres = stats.ttest_ind(action_scores, sports_scores, equal_var=(levene_genres.pvalue > 0.05))

print(f"P-value para géneros: {results_genres.pvalue}")

if results_genres.pvalue < alpha:
    print("Rechazamos la hipótesis nula: las calificaciones promedio por género son diferentes.")
else:
    print("No podemos rechazar la hipótesis nula: no hay evidencia de diferencia.")
````

Explicación:

- Formulación: La hipótesis nula H0 siempre se plantea como una igualdad o ausencia de diferencia. La alternativa H1 es lo que queremos investigar (si son distintas).
- Criterio Estadístico: Se eligió ttest_ind porque estamos comparando las medias de dos grupos independientes.
